# GPT load

```python
from transformers import pipeline

generator = pipeline("text-generation", model="openai-community/gpt2")

prompt = "who are you?"
output = generator(prompt,
                   max_length= 50,
                   num_return_sequences=1,
                   max_new_tokens=50,
                   do_sample=True,
                   temperature=0.7)

print(output[0]['generated_text'])
```

# BERT load

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis",
                      model = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
                      )

text = "I hate learning Machine Learning and I love Deep Learning."

results = classifier(text)

results[0]['label']
```

# T5 load

```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

# Under transformers v5, we load seq2seq models directly instead of using the removed pipelines
model_name = "google-t5/t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = """
Machine learning is a branch of artificial intelligence.
It allows computers to learn patterns from data.
Deep learning is a subset of machine learning.
It uses neural networks with multiple layers.
These technologies are widely used in healthcare,
finance, education and computer vision.
"""

# Prepare input with the T5 prefix
inputs = tokenizer(f"summarize: {text}", return_tensors="pt", max_length=512, truncation=True)

# Generate summary
outputs = model.generate(
    **inputs,
    max_length=60,
    min_length=10,
    num_beams=4,
    early_stopping=True
)

# Decode and print results
summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(summary)
```
